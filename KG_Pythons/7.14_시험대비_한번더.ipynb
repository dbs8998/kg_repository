````python
# ## ✅ 문제 1. (중급 ⭐⭐⭐)

# **[계산기 프로그램 만들기]**

# ✅ 조건:

# - 사용자로부터 숫자 2개와 연산자(+, -, *, /)를 입력받는다.
# - 해당 연산을 수행하고 결과를 출력한다.
# - 0으로 나누는 경우 예외 처리한다.

# ✅ 예시 실행:

# ```
# javascript
# 복사편집
# 첫 번째 숫자를 입력하세요: 10
# 두 번째 숫자를 입력하세요: 2
# 연산자를 입력하세요 (+, -, *, /): /
# 결과: 5.0

# ```

try:
    num1 = int(input('첫 번째 숫자를 입력하세요:'))
    num2 = int(input('두 번째 숫자를 입력하세요:'))

    s = input('연산자를 입력하세요 (+, -, *, /):')

    result = 0
    if s == '+':
        result = num1 + num2
    elif s == '-':
        result = num1 - num2
    elif s == '*':
        result = num1 * num2
    elif s == '/':
        if num1 == 0 or num2 == 0:
            raise ValueError('0으로 나누기를 할 수 없습니다.')
        result = num1 / num2
    else:
        raise ValueError('연산자를 잘못 입력했습니다. :', s)
    print(f"결과값 : {result}")

except ValueError as e:
    print('잘못 입력했습니다.', e)
````

````python
# ## ✅ 문제 2. (중급 ⭐⭐⭐)

# **[랜덤 숫자 맞추기 게임]**

# ✅ 조건:

# - 1~100 사이 난수를 생성
# - 사용자가 맞출 때까지 반복
# - 입력값이 정답보다 크면 "너무 높아요", 작으면 "너무 낮아요" 출력
# - 정답을 맞추면 "축하합니다. 정답입니다." 출력 후 종료
# - 숫자가 아닌 값 입력 시 예외 처리

# ✅ 예시 실행:

# 1부터 100 사이의 숫자를 추측해보세요: 50
# 너무 높아요
# ...
# 축하합니다. 정답입니다.

import random

random_int = random.randint(1,100)

while True:
    try:
        num = int(input('1부터 100 사이의 숫자를 추측해보세요:'))

        if num > 100 or num < 1:
            raise ValueError('1부터 100 사이의 숫자를 입력하세요.')
        elif num > random_int:
            print('너무 높아요')
        elif num < random_int:
            print('너무 낮아요')
        else:
            print('축하합니다. 정답입니다.')
            break

    except ValueError as e:
        print('잘못 입력하셨습니다.',e)
````

````python
# ## ✅ 문제 4. (중급 ⭐⭐⭐)

# **[리스트의 점수 → 등급 변환 함수 작성하기]**

# ✅ 조건:
# - 리스트 형태의 점수(0~100)를 받아서
# - 90점 이상: 'A', 80점 이상: 'B', 70점 이상: 'C', 그 이하: 'F'
# - 변환된 리스트를 반환
# - 함수 이름: `grades()`

def grades(arr):
    grade_arr = []
    for i in arr:
        t = ""
        if i >= 90:
            grade_arr.append('A')
        elif i >= 80 and i < 90:
            grade_arr.append('B')
        elif i >= 70 and i < 80:
            grade_arr.append('C')
        else:
            grade_arr.append('F')
    return grade_arr

try:

    input_arr = input('0~100사이의 점수를 입력하세요. (ex 98 70 58 25):')

    split_arr = input_arr.split()

    chks = []
    for i in split_arr:
        num = int(i)

        if num > 100 or num < 0:
            raise ValueError('0~100사이의 숫자를 입력하세요.', i)
        else:
            chks.append(num)
    
    print(grades(chks))
        

except ValueError as e:
    print('잘못입력하셨습니다.', e)
````

````python
# ## ✅ 문제 12. (중급 ⭐⭐⭐)

# **[상속 클래스에서 별도의 함수 추가 구현]**

# 아래 조건을 만족하는 부모 클래스(Person)와 자식 클래스(Student)를 파이썬으로 구현하세요:

# ✅ 부모 클래스: `Person`

# - 속성: name(이름), age(나이)
# - 생성자에서 name, age 초기화
# - introduce() 메서드
#     - 출력: "저는 [name], [age]살입니다."

# ✅ 자식 클래스: `Student`

# - `Person` 클래스를 상속
# - 추가 속성: major(전공)
# - 생성자에서 name, age, major 초기화 (부모 생성자 호출)
# - study() 메서드
#     - 출력: "[name]이(가) [major]를 공부합니다."

# ✅ 예시 실행:

# ```
# student = Student("Bob", 20, "Computer Science")
# student.introduce()
# student.study()

# 저는 Bob, 20살입니다.
# Bob이(가) Computer Science를 공부합니다.
# ```


class Person():
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print(f'저는 {self.name}, {self.age}살입니다.')

class Student(Person):
    def __init__(self, name, age, major):
        super().__init__(name, age)
        self.major = major

    def study(self):
        print(f'{self.name}이(가) {self.major}를 공부합니다.')

student = Student("Bob", 20, "Computer Science")
student.introduce()
student.study()
````

````python
# ✅ ⭐️ 하급 문제 ③

# # ⭐️ 문제 3
# # [고정 색상 범위로 마스크 생성하기]

# ✅ 조건:
# - HSV 이미지에서 특정 색상 범위(H_min ~ H_max, S_min ~ S_max, V_min ~ V_max)를 지정
# - cv2.inRange()를 사용해 마스크 생성
# - 마스크 출력

# ✅ 목표 함수:
# - cv2.inRange()
# - cv2.imshow()

import cv2
import numpy as np

img = cv2.imread('color_test.jpg')

img = cv2.resize(img,(680,480))

hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

min = np.array([50,80,30])
max = np.array([70,255,255])

mask = cv2.inRange(hsv, min, max)

cv2.imshow('mask',mask)
cv2.imshow('img',img)

cv2.waitKey(0)
cv2.destroyAllWindows()
````

````python

# ✅ ⭐️ 중급 문제 ④

# # ⭐️ 문제 4
# # [마스크로 색상 영역만 추출하기]

# ✅ 조건:
# - inRange로 만든 마스크를 사용해
# - 원본 이미지에서 특정 색상 영역만 컬러로 유지
# - 나머지는 제거(검정색 처리)
# - 결과 출력

# ✅ 목표 함수:
# - cv2.bitwise_and()
# - cv2.imshow()

import cv2
import numpy as np

img = cv2.imread('color_test.jpg')
img = cv2.resize(img, (680,480))

hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

min = np.array([30,30,30])
max = np.array([80,255,255])

mask = cv2.inRange(hsv, min, max)

result = cv2.bitwise_and(img, img, mask=mask)

cv2.imshow('result', result)

cv2.waitKey(0)
cv2.destroyAllWindows()

````

````python
# ✅ ⭐️ 중급 문제 ⑤

# # ⭐️ 문제 5
# # [마스크에서 윤곽선 검출 + 사각형 표시]

# ✅ 조건:
# - cv2.findContours()로 마스크에서 윤곽선 탐색
# - 각 검출된 윤곽선의 외접 사각형(bounding box) 계산
# - 원본 이미지 위에 사각형을 그려 강조
# - 결과 출력

# ✅ 목표 함수:
# - cv2.findContours()
# - cv2.boundingRect()
# - cv2.rectangle()
# - cv2.imshow()

import cv2
import numpy as np

img = cv2.imread('color_test.jpg')
img = cv2.resize(img, (680,480))

hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

min = np.array([30,50,50])
max = np.array([80,255,255])

mask = cv2.inRange(hsv, min, max)

contours, _ =cv2.findContours(mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

 # ⑩ 윤곽선마다 사각형 그리기
for cnt in contours:
    area = cv2.contourArea(cnt)
    if area > 100:  # 너무 작은 잡음은 무시
        x, y, w, h = cv2.boundingRect(cnt)
        cv2.rectangle(img, (x, y), (x + w, y + h), (0, 255, 0), 2)


cv2.imshow('img', img)

cv2.waitKey(0)
cv2.destroyAllWindows()
````
