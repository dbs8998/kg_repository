````python
# ## ✅ 문제 1. (중급 ⭐⭐⭐)

# **[계산기 프로그램 만들기]**

# ✅ 조건:

# - 사용자로부터 숫자 2개와 연산자(+, -, *, /)를 입력받는다.
# - 해당 연산을 수행하고 결과를 출력한다.
# - 0으로 나누는 경우 예외 처리한다.

# ✅ 예시 실행:

# ```
# javascript
# 복사편집
# 첫 번째 숫자를 입력하세요: 10
# 두 번째 숫자를 입력하세요: 2
# 연산자를 입력하세요 (+, -, *, /): /
# 결과: 5.0

# ```

result = 0
while True:
    num1 = int(input('첫 번째 숫자를 입력하세요:'))
    num2 = int(input('두 번째 숫자를 입력하세요:'))

    if num1 == 0 or num2 == 0:
        print('0을 입력하셨습니다. 다시 입력하세요.')
        continue

    a = input('연산자를 입력하세요 (+, -, *, /):')


    if a == '+':
        result = num1 + num2
        break
    elif a == '-':
        result = num1 - num2
        break
    elif a == '*':
        result = num1 * num2    
        break
    elif a == '/':
         result = num1 / num2
         break
    else:
        print('연산자를 잘못 입력하셨습니다.')
        break

print('결과:', result)
````

````python
# ## ✅ 문제 2. (중급 ⭐⭐⭐)

# **[랜덤 숫자 맞추기 게임]**

# ✅ 조건:

# - 1~100 사이 난수를 생성
# - 사용자가 맞출 때까지 반복
# - 입력값이 정답보다 크면 "너무 높아요", 작으면 "너무 낮아요" 출력
# - 정답을 맞추면 "축하합니다. 정답입니다." 출력 후 종료
# - 숫자가 아닌 값 입력 시 예외 처리

# ✅ 예시 실행:

# 1부터 100 사이의 숫자를 추측해보세요: 50
# 너무 높아요
# ...
# 축하합니다. 정답입니다.
import random

val = random.randint(1,100)

while True:
    try:
        num = int(input('1부터 100 사이의 숫자를 추측해보세요:'))

        if val < num:
            print('너무 높아요')
        elif val > num:
            print('너무 낮아요')
        else:
            print('축하합니다. 정답입니다.')
            break

    except ValueError:
        print('숫자를 입력하세요.')


````

````python
# ## ✅ 문제 3. (중급 ⭐⭐⭐)

# **[복권 번호 맞추기 게임]**

# ✅ 조건:

# - 10부터 99까지의 난수(2자리 정수)를 생성
# - 사용자가 맞출 때까지 반복
# - 힌트: 입력값이 정답과 일치하지 않으면 "다시 시도하세요" 출력
# - 맞으면 "당첨! 복권 번호는 [번호]입니다" 출력

# ✅ 예시 실행:
# 복권 번호를 맞춰보세요(10~99): 25
# 다시 시도하세요.
# ...
# 당첨! 복권 번호는 47입니다.

import random

result = random.randint(10,99)

while True:
    try:
        num = int(input('복권 번호를 맞춰보세요(10~99):'))
        if result == num:
            print(f'당첨! 복권 번호는 {result}입니다.')
            break
        else:
           print('다시 시도하세요.') 
    except ValueError:
        print('숫자를 입력하세요.')



````
