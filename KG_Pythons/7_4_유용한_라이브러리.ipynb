```python
import numpy as np

arr1 = np.array([1,2,3])

arr2 = np.array([4,5,6])


#덧셈
#result_add = np.add(arr1, arr2)
result_add = arr1 + arr2
print("덧셈 :",result_add)

#뺄셈
#result_sub = np.subtract(arr1, arr2)
result_sub = arr1-arr2
print("뺄셈 :",result_sub)

#곱셉
#result_mul = np.multiply(arr1, arr2)
result_mul = arr1*arr2
print("곱셈 :",result_mul)

#나눗셈
#result_div = np.divide(arr1, arr2)
result_div = arr1/arr2
print("나눗셈 :",result_div)
```

```python
#브로드 캐스팅 - 브로드캐스팅(Broadcasting)은 크기가 다른 배열 간의 연산을 자동으로 맞춰서 수행하는 규칙
import numpy as np

arr1 = np.array([[1,2,3],[4,5,6]])
arr2 = np.array([1,2,3])

result = arr1 + arr2

print(result)
```

```python
arr = np.array([[1,2,3],[4,5,6]])

#합계
sum_result = np.sum(arr)
print("합계 :",sum_result)

#평균
mean_result = np.mean(arr)
print("평균 :",mean_result)

#표준편차
std_result = np.std(arr)
print("표준편차 :",std_result)

#특정 축을 따라 합계 계산
sum_axis0 = np.sum(arr, axis=0)
print("열별 합계 : ", sum_axis0)

sum_axis1 = np.sum(arr, axis=1)
print("행별 합계 : ", sum_axis1)
```

```python
import numpy as np

arr = np.array([3,2,1,5,4])

#정렬
sorted_arr = np.sort(arr)
print(sorted_arr)

#2차원 배열 정렬
arr2 = np.array([[3,1,2],[6,4,5]])

#행별 정렬
sorted__arr_row = np.sort(arr2, axis=1)
print(sorted__arr_row)

#열별 정렬
sorted_arr_col = np.sort(arr2, axis=0)
print(sorted_arr_col)

```

```python
import numpy as np

#두 행렬 생성
matrix1 = np.array([[1,2],[3,4]])
matrix2 = np.array([[5,6],[7,8]])

#행렬곱셈
matrix_product= np.dot(matrix1, matrix2)
print("행렬 곱셈 결과 : \n",matrix_product)

#전치 행렬
transpose_matrix=np.transpose(matrix1)
print("전치 행렬 결과 : \n",transpose_matrix)


#역행렬
inverse_matrix = np.linalg.inv(matrix1)
print("역행렬 결과 : \n",inverse_matrix)

#행렬식
det_matrix = np.linalg.det(matrix1)
print("행렬식 결과 : \n",det_matrix)
```

```python
# Name   Age   Score
# 아름    24    85
# 다현    23    90
# 준호    22    88
# 민수    19    75
# 서준    10    60
# 이 DataFrame을 생성하고:

# 1️⃣ 나이가 20 이상인 사람만 선택
# 2️⃣ 선택된 사람들의 평균 점수 계산

# 요구 출력 예:

# markdown
# 복사
# 편집
# 나이 20 이상:
#     Name  Age  Score
# 0    아름   24     85
# 1    다현   23     90
# 2    준호   22     88

# 평균 점수: 87.67

import pandas as pd

data = {'Name' : ['아름', '다현','준호','민수','서준'],
        'Age' : [24,23,22,19,10],
        'Score' : [85,90,88,75,60]}

df = pd.DataFrame(data)

filtered_df = df[df['Age'] >= 20]

total =  0
cnt = 0
for n in filtered_df['Score']:
    total += n
    cnt += 1
avg = total / cnt

print(filtered_df)
print(round(avg, 2))
    
```
