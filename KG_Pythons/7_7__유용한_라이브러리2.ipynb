```python
import matplotlib.pyplot as plt
import numpy as np

# 데이터 생성
x = np.linspace(0, 10, 120)  
# 0부터 10까지 120개의 균일한 값 생성

y = 2 + 1 * np.sin(2 * x)    
# 사인 함수 변형: 진폭=1, 주기=pi, y축 이동=+2

x2 = np.linspace(0, 10, 25)  
# 0부터 10까지 25개의 값 (표식용으로 적은 개수)

y2 = 2 + 1 * np.sin(2 * x2)  
# 같은 함수이지만 x2 포인트에 대응하는 y2

# plot
fig, ax = plt.subplots()  
# 새로운 figure와 하나의 axes(좌표축) 생성
# - fig : 전체 그림
# - ax : 실제 그리는 좌표 영역

# 선 + 마커
ax.plot(x2, y2 + 2.5, 'ro-', markeredgewidth=2)
# 빨간색(r) 원형(o) 마커 + 선(-)
# y2를 위로 2.5 이동 (y축 위치 변경)
# markeredgewidth=2 : 마커 테두리 두께 설정

# 부드러운 곡선
ax.plot(x, y, linewidth=2.0)
# 세밀한 x로 부드러운 사인 곡선
# linewidth=2.0 : 선 두께

# 교차 마커만
ax.plot(x2, y2 - 2.5, 'gx', linewidth=2)
# 초록색(g) X 마커
# y2를 아래로 2.5 이동

# 축 범위와 눈금 설정
ax.set(
    xlim=(0, 8),                    # x축 범위 0~8
    xticks=np.arange(0, 8),         # x축 눈금 위치 (0,1,2,...7)
    ylim=(-2, 6),                   # y축 범위 -2~6
    yticks=np.arange(-2, 6)         # y축 눈금 위치 (-2,-1,...5)
)

plt.show()
# 그림 출력
```

```python
import numpy as np
import matplotlib.pyplot as plt


def f(t):
    return np.exp(-t) * np.cos(2*np.pi*t)

t1 = np.arange(0.0,5.0,0.5)
t2 = np.arange(0.0,5.0,0.05)

plt.figure(1)
plt.subplot(211)
plt.plot(t1, f(t1), 'yo', t2, f(t2), 'k')
plt.subplot(212)
plt.plot(t2, np.cos(2*np.pi*t2), 'g--')
plt.show()
```

```python
import threading
import time

def print_numbers():
    for i in range(5):
        print(i)
        time.sleep(1)

#스레드 생성
threed = threading.Thread(target=print_numbers)

#스레드 시작
threed.start()

#메인 스레드에서 작업
for i in range(5,10):
    print(i)
    time.sleep(1)


threed.join()
print("스레드 종료")
```

```python
import threading as th
import time


def print_numbers(thread_name, delay):
    for i in range(5):
        print(f"{thread_name}: {i}")
        time.sleep(delay)


#스레드 생성
threed1 = th.Thread(target=print_numbers, args=("Threed-1", 1))
threed2 = th.Thread(target=print_numbers, args=("Threed-2", 1.5))

#스레드 시작
threed1.start()
threed2.start()

#스레드 종료될때까지 대기
threed1.join()
threed2.join()
print("종료")
```

```python
import threading as th

shared_counter= 0

lock = threading.Lock()

def increment_counter(thread_name):
    global shared_counter
    for _ in range(100000):
        #Lock 획득
        lock.acquire()
        try:
            shared_counter += 1
        finally:
            lock.release()
    print(f"{thread_name} 완료")

#스레드 생성
threed1 = th.Thread(target=increment_counter, args=("Threed-1", ))
threed2 = th.Thread(target=increment_counter, args=("Threed-2", ))

#스레드 시작
threed1.start()
threed2.start()

#스레드 종료될때까지 대기
threed1.join()
threed2.join()
print("최종 카운터 값 :", shared_counter)
```

```python
import threading
import time
import queue

q = queue.Queue()

def producer():
    for i in range(5):
        q.put(i)
        print(f"생산자 : {i} 생성")
        time.sleep(1)

def consumer():
    while True:
        item = q.get()
        if item is None:
            break
        print(f"소비자 : {item} 소비")
        time.sleep(2)

#스레드 생성
producer_thread = threading.Thread(target=producer)
consumer_thread = threading.Thread(target=consumer)

producer_thread.start()
consumer_thread.start()

producer_thread.join()

q.put(None)

consumer_thread.join()
print("모든 스레드 종료")


```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1️⃣ 스크래핑
url = 'https://finance.naver.com/sise/sise_index_day.nhn?code=KOSPI&page=1'
headers = {'User-Agent': 'Mozilla/5.0'}
response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.text, 'html.parser')

# 2️⃣ 테이블 추출
table = soup.select_one('table.type_1')
dfs = pd.read_html(str(table))
df = dfs[0]

# 3️⃣ 컬럼명 클린
df.columns = df.columns.str.strip()
print("컬럼명:", df.columns)

# 4️⃣ 유효 행만
df = df.dropna()

# 5️⃣ 필요한 컬럼만
df = df[['날짜', '체결가']]
df = df.rename(columns={'날짜': 'Date', '체결가': 'Close'})

# 6️⃣ 데이터 변환
df['Date'] = pd.to_datetime(df['Date'], format='%Y.%m.%d')
df['Close'] = df['Close'].astype(str).str.replace(',', '').astype(float)

# 7️⃣ 날짜 순 정렬
df = df.sort_values('Date')
print(df.head())

# 8️⃣ NumPy 변환
dates = df['Date'].to_numpy()
prices = df['Close'].to_numpy()

# 9️⃣ 시각화
plt.figure(figsize=(10, 5))
plt.plot(dates, prices, 'bo-', label='KOSPI Close')
plt.title('KOSPI Closing Price (체결가)')
plt.xlabel('Date')
plt.ylabel('Close Price')
plt.grid(True)
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
