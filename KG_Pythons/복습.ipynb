```python

# 문제 1. 딕셔너리와 리스트를 이용한 재고 관리 시스템
# 아래 조건을 만족하는 재고 관리 프로그램을 만들어라.

# inventory라는 딕셔너리로 제품 이름을 키, 수량을 값으로 관리한다.

# 사용자로부터 계속해서 아래의 명령을 입력받는다:

# add <제품명> <수량> : 재고를 추가

# sell <제품명> <수량> : 재고를 차감. 재고 부족 시 오류 메시지 출력

# list : 현재 재고를 제품명 알파벳순으로 출력

# exit : 프로그램 종료

# 예시
# 명령을 입력하세요: add apple 10
# apple 10개가 추가되었습니다.

# 명령을 입력하세요: sell apple 3
# apple 3개 판매 완료. 남은 수량: 7

# 명령을 입력하세요: list
# apple: 7

# 명령을 입력하세요: sell apple 10
# 재고가 부족합니다.

# 명령을 입력하세요: exit
# 프로그램을 종료합니다.
inventtory = {}


while True:
    try:
        msg = input('명령을 입력하세요:')
        content = msg.split(' ')
        print(content)

        if content[0] == 'add':
            print(f"{content[1]} {content[2]}개가 추가되었습니다.")
            if inventtory.get(content[1]) != None:
                inventtory[content[1]] = int(inventtory.get(content[1])) + int(content[2])
            else:
                inventtory[content[1]] = int(content[2])
        elif content[0] == 'sell':
            if inventtory.get(content[1]) == None:
                print('재고가 부족합니다.')
            else:
                if (int(inventtory.get(content[1])) - int(content[2])) >= 0:
                    inventtory[content[1]] = int(inventtory.get(content[1])) - int(content[2])
                    print(f"{content[1]} {content[2]}개 판매 완료. 남은 수량: {inventtory[content[1]]}")
                else:
                   print('재고가 부족합니다.') 
        elif content[0] == 'list':
            
            for k,v in inventtory.items():
                print(f"{k} : {v}")
        elif content[0] == 'exit':
            print('프로그램을 종료합니다.')
            break
        else:
            print('잘못된 명령어입니다.')
    except Exception as e:
        print('잘못된 명령어 입니다.', e)

    
```

```python
# ✅ 문제 2. 리스트 내포(Comprehension) + 조건
# 아래 리스트가 있다.

# python
# 복사
# 편집
# numbers = [1, 4, 9, 16, 25, 36, 49, 64]
# 리스트 내포를 사용해 다음 규칙을 만족하는 새로운 리스트를 만들어라:

# 각 원소의 제곱근을 구한다.

# 제곱근이 짝수인 수만 남긴다.

# 출력 예시

# csharp
# 복사
# 편집
# [2.0, 4.0, 6.0, 8.0]

numbers = [1, 4, 9, 16, 25, 36, 49, 64]
t = [x ** 0.5 for x in numbers]


result=[]
for i in t:
    if i % 2 == 0:
        result.append(i)
print(result)
```

```python
# ✅ 문제 3. 중첩 딕셔너리 학생관리
# 학생들의 정보가 다음과 같은 딕셔너리 형태로 주어진다.

# python
# 복사
# 편집
# students = {
#     "Alice": {"age": 20, "scores": {"python": 85, "arduino": 90}},
#     "Bob": {"age": 21, "scores": {"python": 78, "arduino": 88}},
#     "Charlie": {"age": 22, "scores": {"python": 92, "arduino": 95}}
# }
# 아래 요구를 만족하는 코드를 작성하라:

# 모든 학생의 이름과 파이썬 점수를 출력

# 파이썬 점수가 80 이상인 학생만 따로 딕셔너리로 추출

# 예시 출력

# yaml
# 복사
# 편집
# Alice: 85
# Bob: 78
# Charlie: 92

# 80점 이상 학생:
# {'Alice': 85, 'Charlie': 92}

students = {
    "Alice": {"age": 20, "scores": {"python": 85, "arduino": 90}},
    "Bob": {"age": 21, "scores": {"python": 78, "arduino": 88}},
    "Charlie": {"age": 22, "scores": {"python": 92, "arduino": 95}}
}

print(students.keys())
hiStudents = {}
for i in students.keys():
    name = i
    scores = students.get(i).get('scores').get('python')
    if int(scores) > 80:
        hiStudents[name] = scores

print(hiStudents)
```

```python
# ✅ 문제 4. 파일 읽고 처리하기
# 아래 조건의 텍스트 파일 **data.txt**가 있다고 가정하라.

# nginx
# 복사
# 편집
# apple
# banana
# cherry
# apple
# orange
# banana
# apple
# 이 파일을 읽어서 다음 작업을 수행하는 코드를 작성하라:

# 각 과일의 등장 횟수를 세어 딕셔너리에 저장

# 알파벳순으로 정렬해 출력

# 예시 출력

# makefile
# 복사
# 편집
# apple: 3
# banana: 2
# cherry: 1
# orange: 1

from pathlib import Path

path = Path('data.txt')
content = path.read_text()

content_list = content.splitlines()

result = {}
for x in content_list:
    if result.get(x) != None:
        result[x] = int(result.get(x)) + 1 
    else:
        result[x] = 1
sorted_fruits = dict(sorted(result.items()))
print(sorted_fruits)
```
